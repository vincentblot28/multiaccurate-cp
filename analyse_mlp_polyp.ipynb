```python
import pickle
```

```python
with open("polyp_recalls_precisions_multiproc.pkl", "rb") as f:
    recalls_precisions = pickle.load(f)
```

```python
recalls_precisions
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
f, ax = plt.subplots(2, 1, figsize=(10, 10))

ax[0].hist(
    recalls_precisions["recalls_resnet"], alpha=.7, bins=20,
    label=f"AA-CRC - avg = {round(np.mean(recalls_precisions['recalls_resnet']), 2)}"
)
ax[0].hist(
    recalls_precisions["recalls_crc"], alpha=.7, bins=20,
    label=f"CRC - avg = {round(np.mean(recalls_precisions['recalls_crc']), 2)}"
)
ax[0].set_title(f"Recall distribution on polyp dataset")
ax[0].legend()

ax[1].hist(
    recalls_precisions["precisions_resnet"], alpha=.7, bins=20,
    label=f"AA-CRC - avg = {round(np.mean(recalls_precisions['precisions_resnet']), 2)}"
)
ax[1].hist(
    recalls_precisions["precisions_crc"], alpha=.7, bins=20,
    label=f"CRC - avg = {round(np.mean(recalls_precisions['precisions_crc']), 2)}"
)
ax[1].set_title(f"Precision distribution on polyp dataset")

ax[1].legend()
plt.show()
```
